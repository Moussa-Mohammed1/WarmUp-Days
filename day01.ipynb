# Day 01 - Python Strings and Lists

Warm-up exercises about **string literals**, **indexing**, **slicing**, and **lists** in Python.

## Raw strings

Put an `r` right before the opening quote. The string is shown **exactly as written** — no special character interpretation (no escape sequences like `\n`, `\t`, or `\` handling).

Useful for Windows paths, regex patterns, etc.

```python
print(r'C:\this\name')
```

## Triple-quoted strings (multi-line literals)

Use `"""..."""` or `'''...'''` to gather multiple lines in one string. A leading backslash `\` right after the opening quotes skips the very first newline.

```python
print("""\
Usage: thingy [OPTIONS]
     -h                        Display this usage message
     -H hostname               Hostname to connect to
""")
```

## Indexing

Strings can be indexed (zero-based). You can access a character by its position.

- `word[i]` → character at position `i`
- `word[-1]` → last character
- `word[-2]` → second-to-last character
- `len(word)` → number of characters

Indexing **out of range** raises an `IndexError`.

```python
word = 'Moussa-Mohammed1'
print(word[7])   # 7th character (position 7)
print(word[-1])  # last character
print(word[-2])  # second-to-last character
print(len(word)) # length = 16
```

```python
# word[66]  # uncomment to trigger IndexError: string index out of range
```

## Slicing

Slice with `string[start:end]` — **start is included**, **end is excluded**.

- `word[:i] + word[i:]` always equals `word`
- Out-of-range slices are handled **silently** (no error)

```python
print(word[:4] + word[4:])  # concatenating slices gives back the whole word
print(word[0:1] + word[7:8])  # concatenation with the + operator -> 'MM'
print(word[7:-1])             # from position 7 up to the last one -> 'Mohammed'
print(word[:6])               # first 6 characters -> 'Moussa'
print(word[:66])              # out-of-range slice -> silently returns everything
print(word[66:])              # start beyond the end -> silently returns ''
```

## Summary - Strings

| Feature | Syntax | Example | Result |
|---|---|---|---|
| Raw string | `r"..."` | `r'C:\\this\\name'` | `C:\this\name` |
| Multi-line | `"""..."""` | — | lines kept |
| Index | `word[i]` | `word[7]` | `M` |
| Last char | `word[-1]` | `word[-1]` | `1` |
| Length | `len(word)` | `len(word)` | `16` |
| Slice | `word[a:b]` | `word[7:-1]` | `Mohammed` |

# Lists

A list is a **collection of elements**. We write it between `[ ]` and separate the elements with commas. A list can hold **different types** at the same time (numbers, text, booleans...).

```python
my_list = [1, 2, 3]
mixed = ['apple', 42, 3.14, True]
print(my_list)
print(mixed)
```

A list is a **box** where we can group **different values**, even of **different types**.

```python
squares = [1, 4, 9, 16, 25]
print(squares)
```

## Indexing and slicing a list

Lists work **like strings**: indexed from 0, negative index from the end, and you can slice them.

- `my_list[0]` → first element
- `mixed[-1]` → last element
- `len(my_list)` → number of elements

```python
print(my_list[0])   # first element
print(mixed[-1])    # last element
print(len(my_list)) # number of elements
print(my_list[:2])  # slice: first two elements
```

## Lists are mutable (you can change them)

Strings **cannot** be changed, but lists **can**. You can give a new value to one element:

```python
my_list[0] = 10
print(my_list)  # [10, 2, 3]
```

## Add and remove elements

- `.append(element)` → add at the end
- `.insert(position, element)` → add at a specific position
- `.remove(element)` → remove by value
- `.pop()` → remove **and return** the last element

```python
my_list.append(4)
print(my_list)  # add at the end -> [10, 2, 3, 4]

my_list.insert(1, 99)
print(my_list)  # add at position 1 -> [10, 99, 2, 3, 4]

my_list.remove(99)
print(my_list)  # remove the value 99 -> [10, 2, 3, 4]

last = my_list.pop()
print(last)     # pop gives back the last element -> 4
print(my_list)  # now -> [10, 2, 3]
```

## Check if an element exists

Use the `in` keyword. The result is `True` or `False`.

```python
print(2 in my_list)   # is 2 inside my_list? -> True
print(42 in my_list)  # is 42 inside my_list? -> False
```
