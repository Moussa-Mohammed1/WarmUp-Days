<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 1 : Présentation personnalisé</span></H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objectif</span> : </br>
Réutiliser un programme existant (Challenge 2) en le structurant dans une fonction nommée horaire_sup.

2. <span style="font-weight: bold; color: #e56922ff;">Travail à faire</span>:</br>
Réécris le programme du Challenge 2 dans une fonction appelée horaire_sup.

```python
import sys
def horaire_sup(nom, nbr_hours, salaire_horaire):
    salaire_total = 0
    if nbr_hours > 40:
        salaire_total = (40 * salaire_horaire) + (nbr_hours - 40) * (salaire_horaire * 1.5)
    else:
        salaire_total = nbr_hours * salaire_horaire
    print(f"Nom d'utilisateur: {nom}. Salaire totale calculé: {salaire_total}")
    
nom = input("Etrer votre nom: ")
try: 
    nbr_hours = float(input("Entrer le nombre des heures que vous avez travaillé: "))
    salaire_horaire = float(input("Entrer votre salaire horaire: "))
    horaire_sup(nom, nbr_hours, salaire_horaire)
except ValueError:
    print("Le salaire horaire et le nombre des heures travaillé doit etre decimal")
    sys.exit()



```

<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 2 : Fonction calculation() – somme et différence</span></H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objectif</span> : </br>
Apprendre à écrire une fonction avec plusieurs paramètres et plusieurs valeurs de retour.

2. <span style="font-weight: bold; color: #e56922ff;">Travail à faire</span>:</br>
Écris une fonction calculation() qui prend deux arguments et calcule leur somme ainsi que leur différence. La fonction doit ensuite retourner les deux résultats dans une seule instruction return.

```python
def calculations(number1, number2):
    somme = number1 + number2
    difference = number1 - number2
    return number1 + number2, difference

number1 = float(input("Entrer le premier nombre: "))
number2 = float(input("Entrer le deuxieme nombre: "))
somme, difference = calculations(number1, number2)
print("Somme: ", somme)
print("Difference: ", difference)
```
