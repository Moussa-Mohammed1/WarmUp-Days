# Day 02 - Loops: `for` and `while`

A **loop** repeats the same code many times. In Python there are two kinds of loops: `for` and `while`.

## `for` loop over a list

`for` goes through each element of a list, one by one. Every round, the element is put inside the variable `i`.

````python
for i in [1, 2, 4, 9, 3]:
    print(i, "\n")
````

## `range()`

`range()` gives a series of numbers, and works with one, two or three arguments:

- `range(7)` → 0, 1, 2, 3, 4, 5, 6
- `range(start, end)` → **start is included**, **end is excluded**
- `range(start, end, step)` → adds a step

````python
for i in range(7):  # 7 numbers: 0, 1, 2, 3, 4, 5, 6
    print(i, "\n")
````

## Loop with an index

Combine `range(len(list))` to get each **position** `i`, then read the element with `list[i]`.

You can also simply loop over the list directly with `for i in list`.

Look at both ways below:

````python
c = ["Full", "-", "Stack", "Developer"]
print(len(c))

# way 1: loop over the positions (indexes)
for i in range(len(c)):
    print("i = ", i, ", c[i] = ", c[i])
print("ENDFOR i in range(len(c))\n")

# way 2: loop over the values directly
for i in c:
    print(i)
print("ENDFOR i in c\n")
````

## `while` loop

`while` repeats **as long as** the condition is `True`.

Careful: you must change the variable inside the loop, or it will run forever!

````python
x = 1
while x < 10:
    print("x = ", x)
    x += 2   # x grows: 1, 3, 5, 7, 9, then 11 (stop)
print("ENDWHILE\n")
````

## `while` + index over a list

Same idea as `for i in range(len(c))`, but with a `while` and a counter `i` we increase by hand.

````python
i = 0
while i < len(c):
    print(c[i])
    i += 1
print("ENDWHILE i < len(c)")
````

## `break` — stop the loop early

`break` exits the loop immediately. Here `while True` would run forever, but we stop it when we find the element `"-"`.

````python
j = 0
while True:            # infinite loop...
    print(c[j])
    if c[j] == "-":   # ...stopped by break
        break
    j += 1
print("ENDWHILE True")
````

## Summary - Loops

| Loop | What it does |
|---|---|
| `for i in [1, 2, 3]` | repeat once for each value |
| `for i in range(n)` | repeat `n` times, with `i` = 0, 1, ..., n-1 |
| `for i in range(len(c))` | loop over positions, read `c[i]` |
| `while condition:` | repeat while condition is `True` |
| `while True:` + `break` | infinite loop stopped by `break` |
| `x += 2` | same as `x = x + 2` |

# Functions

A function is a **block of code** that we can **call** (run) many times.

It takes **inputs**, does something, and can give back a **result**.

## The anatomy of a function (part by part)

Look at each part of this function:

```python
def add(a, b):
    return a + b
```

| Part | What it is |
|---|---|
| `def` | keyword that CREATES the function |
| `add` | the NAME of the function (you choose it) |
| `(a, b)` | the PARAMETERS: the inputs |
| `:` | the header ends with a colon |
| body (indented) | the code that runs when we call it |
| `return a + b` | gives BACK the result |

## Call the function and use the result

To call a function: write its name followed by `( )` with the inputs. The result of `return` can be stored in a variable.

````python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)       # 8
print(add(10, 20))  # 30
````

## A function without parameters

If the function needs no input, the `( )` stays empty. We can call it as many times as we want.

````python
def greet():
    print("Hello!")

greet()
greet()  # we can call it as many times as we want
````

## A function with one parameter

The input is used inside the body — each call can send a different value.

````python
def greet_user(name):
    print("Hello", name)

greet_user("Youco")
greet_user("Moussa")
````

## A function that returns a value

`return` gives back a result. We can store it or use it directly in a `print`.

````python
def multiply(a, b):
    return a * b

total = multiply(4, 5)
print(total)          # 20
print(multiply(2, 3)) # 6
````

## Labelled parameters (keyword arguments)

When we call a function, we can write `parameter=value`. Then the **order does not matter** anymore.

````python
def greet_full(first, last):
    print("Hello", first, last)

greet_full(first="Moussa", last="Youco")
greet_full(last="Youco", first="Moussa")  # same result, order changed
````

## Default values

Write `=value` inside the `def`. Then we can call the function **without** that argument, and it uses the default.

````python
def greet(name="guest"):
    print("Hello", name)

greet()          # uses the default -> Hello guest
greet("Youco")   # no default here -> Hello Youco
````

## Summary - Functions

| Part | Syntax |
|---|---|
| Define | `def name(parameters):` |
| Parameters | the inputs, passed with `(` `)` |
| Body | indented code, runs when called |
| Return | `return value` gives back a result |
| Call | `name(args)` runs the function |
| Keyword arg | `name(param=value)` order doesn't matter |
| Default value | `def f(name="guest")` call without it
