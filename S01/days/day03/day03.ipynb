# Day 03 - Data Structures

A **data structure** is a way to **group data together**.

Python has: lists `[ ]`, tuples `( )`, dicts `{ }`, sets `{ }`.

## `tuple` — cannot be changed

A tuple is **like a list**, but it **CANNOT be changed** (immutable). We write it with `( )` instead of `[ ]`.

Good for data that should stay the same (e.g. a point `(x, y)`).

```python
point = (3, 4)
print(point)       # (3, 4)
print(point[0])    # 3
print(len(point))  # 2

# unpack a tuple into variables
x, y = point
print(x, y)        # 3 4
```

```python
point[0] = 10  # uncomment to trigger TypeError: 'tuple' object does not support item assignment
```

## `dict` — key → value pairs

A dict stores **`key` → `value`** pairs. We write it with `{ }` and separate the key from the value with `:`.

- read a value with `person["key"]`
- add or change a pair with `person["new_key"] = value`
- check a key with `"key" in person`

```python
person = {
    "name": "Youco",
    "age": 25,
    "skills": ["Python", "HTML"]
}

print(person["name"])   # Youco
print(person["age"])    # 25

person["city"] = "Oran"   # add a new pair
print(person)              # now with 'city'

print("name" in person)    # does the key exist? -> True

for key in person:          # loop over the keys
    print(key, "->", person[key])
```

## `set` — unique elements, no order

A set is a collection of **UNIQUE elements** with **NO order**.

- duplicates are removed automatically
- the order can change every time (don't rely on it)
- `add()` adds, `remove()` removes

```python
fruits = {"apple", "banana", "apple", "orange"}
print(fruits)                # "apple" appears only ONCE

print("apple" in fruits)     # True

fruits.add("kiwi")           # add an element
print(fruits)

fruits.remove("banana")      # remove an element
print(fruits)
```

## Summary

| Structure | Written with | Can change? | Order? | Duplicates? |
|---|---|---|---|---|
| `list` | `[ ]` | yes | yes | yes |
| `tuple` | `( )` | **no** | yes | yes |
| `dict` | `{ }` | yes | yes (insertion) | keys must be unique |
| `set` | `{ }` | yes | **no** | **no** |
